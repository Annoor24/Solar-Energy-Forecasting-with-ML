```python
# ============================================================
# STAGE 1 - CELL 2: Define Project Folder Paths
# ============================================================

import os

PROJECT_ROOT  = "C:/Users/Mahir/Desktop/Solar_Energy_Forecasting_Project"

DATA_DIR      = os.path.join(PROJECT_ROOT, "data")
KANO_DIR      = os.path.join(DATA_DIR, "Kano")
MAIDUGURI_DIR = os.path.join(DATA_DIR, "Maiduguri")
ABUJA_DIR     = os.path.join(DATA_DIR, "Abuja")
MODELS_DIR    = os.path.join(PROJECT_ROOT, "models")
RESULTS_DIR   = os.path.join(PROJECT_ROOT, "results")
PLOTS_DIR     = os.path.join(PROJECT_ROOT, "plots")

for folder in [MODELS_DIR, RESULTS_DIR, PLOTS_DIR]:
    os.makedirs(folder, exist_ok=True)

print("=" * 50)
print("PROJECT FOLDER STRUCTURE")
print("=" * 50)
print(f"Project Root : {PROJECT_ROOT}")
print(f"Kano folder  : {KANO_DIR}")
print(f"Maiduguri    : {MAIDUGURI_DIR}")
print(f"Abuja folder : {ABUJA_DIR}")
print("=" * 50)
print("Paths defined successfully!")
```

```python
# ============================================================
# STAGE 1 - CELL 3: Verify All 24 Data Files Are Present
# ============================================================

years = list(range(2016, 2024))

# Capital first letter + .csv to match your exact filenames
cities = {
    'Kano'     : KANO_DIR,
    'Maiduguri': MAIDUGURI_DIR,
    'Abuja'    : ABUJA_DIR
}

print("=" * 50)
print("DATA FILE VERIFICATION")
print("=" * 50)

total_files_expected = 0
total_files_found    = 0
missing_files        = []

for city_name, city_path in cities.items():
    print(f"\nChecking {city_name.upper()} folder...")
    print(f"Path: {city_path}")

    for year in years:
        total_files_expected += 1

        # Matches exactly: Kano_2016.csv, Maiduguri_2016.csv, Abuja_2016.csv
        filename = f"{city_name}_{year}.csv"
        filepath = os.path.join(city_path, filename)

        if os.path.exists(filepath):
            size_kb = os.path.getsize(filepath) / 1024
            print(f"  ✅ {filename} found — {size_kb:.1f} KB")
            total_files_found += 1
        else:
            print(f"  ❌ {filename} MISSING!")
            missing_files.append(filename)

print("\n" + "=" * 50)
print("SUMMARY")
print("=" * 50)
print(f"Total files expected : {total_files_expected}")
print(f"Total files found    : {total_files_found}")
print(f"Total files missing  : {total_files_expected - total_files_found}")

if len(missing_files) == 0:
    print("\n🎉 ALL 24 FILES PRESENT! Ready to proceed to Stage 2.")
else:
    print(f"\n⚠️  MISSING FILES:")
    for f in missing_files:
        print(f"   {f}")
```

```python
# ============================================================
# STAGE 1 - CELL 4: Quick Data Peek — Confirm File Structure
# ============================================================

# Load one file to confirm the structure looks correct
# We use skiprows=14 because the first 14 lines are NASA header text
# not actual data — as you saw when we opened the file earlier

sample_file = os.path.join(KANO_DIR, "kano_2016.csv")

df_sample = pd.read_csv(sample_file, skiprows=14)

print("=" * 50)
print("SAMPLE DATA PEEK — KANO 2016")
print("=" * 50)

print(f"\nShape: {df_sample.shape[0]} rows × {df_sample.shape[1]} columns")
print(f"\nColumn names:")
for col in df_sample.columns:
    print(f"  - {col}")

print(f"\nFirst 5 rows:")
print(df_sample.head())

print(f"\nLast 5 rows:")
print(df_sample.tail())

print(f"\nData types:")
print(df_sample.dtypes)

print(f"\nBasic statistics:")
print(df_sample.describe().round(2))
```

```python
# ============================================================
# STAGE 1 - CELL 5: Check -999 Missing Value Pattern
# ============================================================

# NASA uses -999 to represent missing or unavailable data
# We need to understand how many of these exist before Stage 2

print("=" * 50)
print("MISSING VALUE ANALYSIS — KANO 2016")
print("=" * 50)

# Count -999 values in each column
for col in df_sample.columns:
    count_999 = (df_sample[col] == -999.0).sum()
    total     = len(df_sample)
    pct       = (count_999 / total) * 100
    if count_999 > 0:
        print(f"{col}: {count_999} missing values ({pct:.1f}% of data)")

print("\nNote: -999 values in ALLSKY_KT (Clearness Index) are")
print("nighttime hours when clearness index cannot be calculated.")
print("This is physically correct and expected — NOT an error.")

# Confirm what hours have -999
print(f"\nSample of rows with -999 in ALLSKY_KT:")
mask = df_sample['ALLSKY_KT'] == -999.0
print(df_sample[mask][['YEAR','MO','DY','HR','ALLSKY_SFC_SW_DWN','ALLSKY_KT']].head(10))
```
